Bowling Stats 

```python
match_bowler_stats = deliveries.groupby(
    ['match_id','season','bowler',
     'batting_team','bowling_team',
     'venue']
).agg({

'runs_off_bat':'sum',
'ball':'count',
'player_dismissed':'count',
'wides':'sum',
'noballs':'sum'

}).reset_index()
```

```python
match_bowler_stats.columns=[

'match_id',
'season',
'bowler',
'batting_team',
'bowling_team',
'venue',
'runs_conceded',
'balls',
'wickets',
'wides',
'noballs'

]
```

```python
match_bowler_stats['runs_conceded']=(
match_bowler_stats['runs_conceded']
+
match_bowler_stats['wides'].fillna(0)
+
match_bowler_stats['noballs'].fillna(0)
)
```

```python
match_bowler_stats['overs'] = (
    (match_bowler_stats['balls']//6)
    +
    ((match_bowler_stats['balls']%6)/10)
)

match_bowler_stats['SR'] = round(
    match_bowler_stats['balls']
    /
    match_bowler_stats['wickets'],2
)

match_bowler_stats['average'] = round(
    match_bowler_stats['runs_conceded']
    /
    match_bowler_stats['wickets'],2
)

match_bowler_stats = match_bowler_stats.replace(
    [float('inf')],
    0
)

match_bowler_stats['figures'] = (
    match_bowler_stats['wickets']
    .astype(int)
    .astype(str)
    +
    "/"
    +
    match_bowler_stats['runs_conceded']
    .astype(int)
    .astype(str)
)

match_bowler_stats.to_csv(
    "match_bowler_stats.csv",
    index=False
)
```

Season Wise Bowler Stats 

```python
season_bowler_stats = match_bowler_stats.groupby(
    ['season','bowler']
).agg({

    'runs_conceded':'sum',
    'balls':'sum',
    'wickets':'sum'

}).reset_index()
```

```python
# overs
season_bowler_stats['overs']=(
    (season_bowler_stats['balls']//6)
    +
    ((season_bowler_stats['balls']%6)/10)
)

# economy
season_bowler_stats['economy']=round(
    season_bowler_stats['runs_conceded']
    /
    season_bowler_stats['overs'],
    2
)

# SR
season_bowler_stats['SR']=round(
    season_bowler_stats['balls']
    /
    season_bowler_stats['wickets'],
    2
)

# average
season_bowler_stats['average']=round(
    season_bowler_stats['runs_conceded']
    /
    season_bowler_stats['wickets'],
    2
)

season_bowler_stats=season_bowler_stats.replace(
    [float('inf')],
    0
)
```

```python
season_bowler_stats.to_csv(
    "season_bowler_stats.csv",
    index=False
)
```
